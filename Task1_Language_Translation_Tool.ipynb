# Task 1: Language Translation Tool

## Description
This project translates text from one language to another using Python and Google Translate (googletrans library).

## Steps:
1. User enters text
2. User selects source and target languages
3. System translates text using API
4. Output is displayed

```python
!pip install googletrans==4.0.0-rc1
```

```python
from googletrans import Translator
```

```python
translator = Translator()
```

```python
text = input("Enter text to translate: ")
```

```python
src_lang = input("Enter source language code (e.g., en, hi, te): ")
dest_lang = input("Enter target language code (e.g., en, hi, te): ")
```

```python
result = translator.translate(text, src=src_lang, dest=dest_lang)
```

```python
print("\nOriginal Text:", text)
print("Translated Text:", result.text)
```

```python
!pip install gTTS
from gtts import gTTS
from IPython.display import Audio
```

```python
tts = gTTS(result.text, lang=dest_lang)
tts.save("output.mp3")

Audio("output.mp3")
```
